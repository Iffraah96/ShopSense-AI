## Load Dataset

```python
df = pd.read_csv('/content/Womens Clothing E-Commerce Reviews.csv')
df.head()
```

```python
#checking duplicate rows
df.duplicated().sum()
```

```python
#Cleaning column names

# Removing extra spaces from column name
df.columns = df.columns.str.strip()

# Replacing space in columns name with underscore
df.columns = df.columns.str.replace(' ','_')

# Coverting column names into lower case
df.columns = df.columns.str.lower()
```

```python
# Display cleaned column names
df.columns
```

## Handling Missing Values

```python
# Checking missing values
df.isnull().sum()/100
```

```python
# dropping column unnamed:_0
# This is an index column and does'nt contribute in the analysis

df.drop(columns=['unnamed:_0'], inplace=True)
```

```python
df.head()
```

```python
#fill the null values in column 'title' with 'No title'
df["title"] = df["title"].fillna("No Title")
```

```python
df.shape
```

```python
# Remove Rows with Missing/null values in column 'Review Text'
df = df.dropna(subset=['review_text'])

# Check the new dataset shape
print("Updated Dataset Shape:", df.shape)
```

```python
# Replacing missing values in following columns with 'unknown'

df['division_name'] = df['division_name'].fillna('Unknown')
df['department_name'] = df['department_name'].fillna('Unknown')
df['class_name'] = df['class_name'].fillna('Unknown')
```

```python
# Verify there are no missing values left
df.isnull().sum()
```

```python
# Save the clean dataset

df.to_csv('clean_women_clothing_review_dataset.csv', index=False)

print('Cleaned dataset saved successfully..!')
```
