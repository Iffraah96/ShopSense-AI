# Notebook 4: Feature Engineering

## Objective

In this notebook, we will convert NLP preprocessed customer review text into numerical features that can be used by machine learning algorithms.

We will use TF-IDF (Term Frequency-Inverse Document Frequency) to represent important words and phrases numerically.

### Steps

1. Load the processed dataset
2. Define the input and target variables
3. Split the data into training and testing sets
4. Create a TF-IDF vectorizer
5. Convert review text into numerical features
6. Analyze important words and phrases
7. Prepare the features for machine learning

The resulting features will be used in the machine learning classification notebook.

## Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 100)

print("Libraries imported successfully!")
```

## Load Dataset

```python
df = pd.read_csv('/content/nlp_preprocessed_women_clothing_review_dataset.csv')
print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("column:", df.dtypes)

df.head()
```

## Select Text and Target Columns

```python
# Seperating text and target variable from the nl preprocessed dataset

# Set text variable (X) as the column final_cleaned_review
X = df['final_cleaned_review']

# Set Target variable (Y) as recommended_ind
# where 1 = recommended and 0 = not recommended
Y = df['recommended_ind']

# Check the number of reviews and target values
print("Number of reviews:", len(X))
print("\nTarget value counts:")
print(Y.value_counts())

# Check for missing values
print("\nMissing values:")
print("Reviews:", X.isna().sum())
print("Target:", Y.isna().sum())
```

## Test Train Split

```python
# Split the data into 80% train and 20% test data
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

# Check the size of X train and test
print('Training reviews:', X_train.shape)
print('Testing reviews:', X_test.shape)

# Check the Y (target) distribution in each dataset
print('\nTraining target distribution:')
print(Y_train.value_counts())

print('\nTest target distribution:')
print(Y_test.value_counts())
```

## Create the TF-IDF Vectorizer

```python
# Create the TF-IDF vectorizer
tfidf = TfidfVectorizer(
    max_features=5000,      # Keep the 5,000 most useful terms
    ngram_range=(1, 2),     # Use single words and two-word phrases
    min_df=2,               # Ignore terms appearing in only 1 review
    max_df=0.95             # Ignore extremely common terms
)

print("TF-IDF vectorizer created successfully!")
```

## Fit and Transform TF-IDF

```python
# Learn the vocabulary and TF-IDF weights from the training reviews
X_train_tfidf = tfidf.fit_transform(X_train)

# Use the same TF-IDF (i.e. X_train_tfidf) vocabulary to transform the testing reviews
X_test_tfidf = tfidf.transform(X_test)

# Print the size of training and test review after tfidf vectorization
print('Training tfidf reviews:', X_train_tfidf.shape)
print('Test tfidf reviews:', X_test_tfidf.shape)
```

## View TF-IDF Feature Names

```python
# Get the words and phrases learned by the TF-IDF vectorizer
feature_name = tfidf.get_feature_names_out()

#Print first 30 tfidf feature names
print('Number of tfidf features: ', len(feature_name))
print('\nList of first 30 features: ')
print(feature_name[:30])
```

```python
# Select the first review from the training data
review_index = 0

# Get the TF-IDF values for this review
review_tfidf = X_train_tfidf[review_index].toarray().flatten()

# Create a table containing each feature and its TF-IDF score
tfidf_scores = pd.DataFrame({
    "feature": feature_name,
    "tfidf_score": review_tfidf
})

# Keep only features that appear in this review
tfidf_scores = tfidf_scores[tfidf_scores["tfidf_score"] > 0]

# Sort from highest to lowest TF-IDF score
tfidf_scores = tfidf_scores.sort_values(
    by="tfidf_score",
    ascending=False
)

# Display the top 15 features for this review
print("Original review:")
print(X_train.iloc[review_index])

print("\nTop TF-IDF features in this review:")
display(tfidf_scores.head(15))
```

## Find the Most Common TF-IDF Features Overall

```python
# Calculate the average TF-IDF score for each feature
average_tfidf = np.asarray(X_train_tfidf.mean(axis=0)).flatten()

# Create a table containing each feature and its average TF-IDF score
top_features = pd.DataFrame({
    "feature": feature_name,
    "average_tfidf": average_tfidf
})

# Sort features from highest to lowest average TF-IDF score
top_features = top_features.sort_values(by="average_tfidf", ascending=False)

# Display the top 20 features
print("Top 20 TF-IDF features across the training dataset:")
display(top_features.head(20))
```

## Save the TF-IDF Vectorizer

```python
# Save the fitted TF-IDF vectorizer
import joblib

joblib.dump(tfidf, "shopsense_tfidf_vectorizer.pkl")

print("TF-IDF vectorizer saved successfully!")
```
